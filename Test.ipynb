```python
import pyalps
import pyalps.plot
import numpy as np
import matplotlib.pyplot as plt
import copy
import math

from quant_rotor.Hamiltonian_models.Dense.heisenberg_hamiltonian import hamiltonian_dense
```

```python
site = 8
state = 2

J = 1
h = -1

J_x = J*1/4
J_y = J*1/4
J_z = J*1/4

h_x = 0
h_y = 0
h_z = h*1/2

g = 1
```

```python
# --- 1) Run ALPS for a specific model ---
parms = [{
    "LATTICE": "chain lattice",
    "MODEL": "spin",
    "local_S": 0.5,
    "L": site,
    "J": J,
    # "Jz": 0.0,
    "h": h,
}]

prefix = "heis_ed"
inp = pyalps.writeInputFiles(prefix, parms)
pyalps.runApplication("fulldiag", inp)  # or "sparsediag"

# --- 2) Load eigenvalues (spectra) ---
files = pyalps.getResultFiles(prefix=prefix)
spec = pyalps.loadSpectra(files)  # energies are in DataSet.y per sector  [oai_citation:1‡ALPS](https://alps.comp-phys.org/documentation/api/loadat/)

# spec is: [ file ][ sector ] -> DataSet, with energies in .y
E_alps = np.sort(np.concatenate([np.asarray(ds.y).ravel() for ds in spec[0]]))

print("First 10 ALPS eigenvalues:", E_alps[:10])
print("Number of eigenvalues loaded:", E_alps.size)
```

```python
H_my, _, _ = hamiltonian_dense(site, J_x, J_y, J_z, h_x, h_y, h_z, g, periodic=True)
```

```python
E_my = np.sort(np.linalg.eigvalsh(H_my))
print("max |ΔE| =", np.max(np.abs(E_my - E_alps)))
```
