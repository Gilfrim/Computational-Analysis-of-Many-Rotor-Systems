## Functions

```python
def h_term(h_upper, h_lower):
    a_h_shift = [i if a_check == a else 0 for a_check in (h_upper, h_lower)]
    return K[a_h_shift[0]:h_upper + a_h_shift[0], a_h_shift[1]:h_lower + a_h_shift[1]]
```

```python
def v_func(x: int, y: int, idx_up_1: int, idx_up_2: int, idx_low_1: int, idx_low_2: int):
    a_v_shift = [
            i if a_check == a else 0
            for a_check in (idx_up_1, idx_up_2, idx_low_1, idx_low_2)
        ]
    if (x - y) == 1:

        return V_yx[
            a_v_shift[0] : idx_up_1 + a_v_shift[0],
            a_v_shift[1] : idx_up_2 + a_v_shift[1],
            a_v_shift[2] : idx_low_1 + a_v_shift[2],
            a_v_shift[3] : idx_low_2 + a_v_shift[3],
        ]

    if (y - x) == 1:

        return V_xy[
            a_v_shift[0] : idx_up_1 + a_v_shift[0],
            a_v_shift[1] : idx_up_2 + a_v_shift[1],
            a_v_shift[2] : idx_low_1 + a_v_shift[2],
            a_v_shift[3] : idx_low_2 + a_v_shift[3],
        ]

    else:
        return np.zeros((idx_up_1, idx_up_2, idx_low_1, idx_low_2), dtype=complex)
```

```python
def A_term(a_upper, a_site):
    return np.hstack((-t_1[a_site], np.identity(a_upper)))

def B_term(b_lower, b_site):
    return np.vstack((np.identity(b_lower), t_1[b_site]))
```

```python
def t_1_func(x: int):
    return t_1[x]
```

```python
def t_2_func(x: int, y: int):
    return t_2[x, y]
```

```python
def ec1(x: int):
    return np.einsum("ia, ai", h_term(i, a), t_1_func(x))
```

```python
def ec2(x: int, y: int):

    term = t_2_func(x, y) + np.einsum("ai, bj-> abij", t_1_func(x), t_1_func(y))

    return np.einsum("ijab, abij", v_func(x, y, i, i, a, a), term)
```

```python
def et1(x: int):
    term = ec1(x)

    for z in range(site):
        if x != z:
            term += ec2(x, z)

    return term
```

```python
def E_tot():
    for x in range(site):
        term = ec1(x)

        for y in range(site):
            if x != y:
                term += ec2(x, y) / 2
```

```python
def c1(x: int):
    return np.einsum("aq, qi -> ai", h_term(a, p), B_term(i, x))
```

```python
def c2(x: int, z: int):
    term = np.einsum("alcd, cdil -> ai", v_func(x, z, a, i, a, a), t_2_func(x, z))
    term += np.einsum("alqs, qi, sl -> ai", v_func(x, z, a, i, p, p), B_term(i, x), B_term(i, z))
    return term
```

```python
def c3(x: int, z: int, w: int):
    return np.einsum("acik, klcp, pl -> ai", t_2_func(x, w), v_func(w, z, i, i, a, p), B_term(i, z))
```

```python
def R_1(x: int):
    term = c1(x)
    for z in range(site):
        if z != x:
            term += c2(x, z)
            for w in range(site):
                if (w != x) and (w != z):
                    term += c3(x, z, w) - et1(x) * t_1_func(x)
    return term
```

```python
def R_2_sym(x: int, y: int):
    term = np.einsum("abrs, ri, sj -> abij", v_func(x, y, a, a, p, p), B_term(i, x), B_term(i, y))
    term += np.einsum("abcd, cdij -> abij", v_func(x, y, a, a, a, a), t_2_func(x, y))
    for z in range(site):
        for w in range(site):
            if z not in {x, y} and w not in {x, y} and z != w:
                term += np.einsum("klcd, acik, bdjl -> abij", v_func(z, w, i, i, a, a) ,t_2_func(y, w))

    term -= t_2_func(x, y) * (et1(x) + et1(y) - ec2(x, y))
    term += np.einsum("ai, bj -> abij", t_1_func(x), t_1_func(y)) * ec2(x, y)

    return term
```

```python
def R_2_non_sym(x: int, y: int):

    term_intermediet = np.zeros((a, i), dtype=complex)

    term = np.einsum("ac, cbij -> abij", h_term(a, a), t_2_func(x, y))

    for z in range(site):
        if (z != x) and (z != y):
            term += np.einsum("acik, kbcs, sj -> abij", t_2_func(x, z), v_func(z, y, i, a, a, p), B_term(i, y))
            term += np.einsum("blds, adij, sl -> abij", v_func(y, z, a, a, i, i), t_2_func(x, y), B_term(i, z))
            term_intermediet += c3(x, y, z) + c3(x, z, y)

    term -= np.einsum("ai, bj -> abij", (c2(x, y) + term_intermediet), t_1_func(y))
    return term
```

```python
def R_2_total(x: int, y: int):
    return R_2_sym(x, y) + R_2_non_sym(x, y) + R_2_non_sym(y, x)
```

```python
def A_term_big(a_site):
    vec = np.vstack((np.identity(i), -t_1[a_site]))

    return np.hstack((vec, np.identity(p)[:, 1:]))

def B_term_big(b_site):
    vec = np.vstack((np.identity(i), t_1[b_site]))
    return np.hstack((vec, np.identity(p)[:, 1:]))
```

```python
def h_bar_term(x: int, h_upper: int, h_lower: int):
    K_bar = np.einsum("ps, sq, qr -> pr", A_term_big(x), h_term(p, p),  B_term_big(x))
    K_bar -= np.eye(p) * K_bar[0,0]

    a_h_shift = [i if a_check == a else 0 for a_check in (h_upper, h_lower)]
    return K_bar[a_h_shift[0]:h_upper + a_h_shift[0], a_h_shift[1]:h_lower + a_h_shift[1]]
```

```python
def v_bar_term(x: int, y: int, idx_up_1: int, idx_up_2: int, idx_low_1: int, idx_low_2: int):

    V_xy_bar = np.einsum("pP, qQ, PQRS, Rr, Ss -> pqrs", A_term_big(x), A_term_big(y), v_func(x, y, p, p, p, p),  B_term_big(x), B_term_big(y))
    V_yx_bar = np.einsum("pP, qQ, PQRS, Rr, Ss -> pqrs", A_term_big(y), A_term_big(x), v_func(y, x, p, p, p, p),  B_term_big(y), B_term_big(x))
    for indx_1 in range(p):
        for indx_2 in range(p):
            V_xy_bar[indx_1, indx_2, indx_1, indx_2] -= V_xy_bar[0, 0, 0, 0]
            V_yx_bar[indx_1, indx_2, indx_1, indx_2] -= V_yx_bar[0, 0, 0, 0]

    a_v_shift = [
            i if a_check == a else 0
            for a_check in (idx_up_1, idx_up_2, idx_low_1, idx_low_2)
        ]
    if (x - y) == 1:

        return V_yx_bar[
            a_v_shift[0] : idx_up_1 + a_v_shift[0],
            a_v_shift[1] : idx_up_2 + a_v_shift[1],
            a_v_shift[2] : idx_low_1 + a_v_shift[2],
            a_v_shift[3] : idx_low_2 + a_v_shift[3],
        ]

    if (y - x) == 1:

        return V_xy_bar[
            a_v_shift[0] : idx_up_1 + a_v_shift[0],
            a_v_shift[1] : idx_up_2 + a_v_shift[1],
            a_v_shift[2] : idx_low_1 + a_v_shift[2],
            a_v_shift[3] : idx_low_2 + a_v_shift[3],
        ]

    else:
        return np.zeros((idx_up_1, idx_up_2, idx_low_1, idx_low_2), dtype=complex)
```

```python
def e_bar_c(x: int, y: int):
    return np.einsum("ijab, abij", v_bar_term(x, y, i, i, a, a), t_2_func(x, y))
```

```python
def E_c(x: int):
    term = 0
    for z in range(site):
        if x != z:
            term += e_bar_c(x, z)
    return term
```

```python
def R_bar_1(x: int):
    term = h_bar_term(x, a, i)
    for z in range(site):
        if x != z:
            term += np.einsum("alcd, cdil -> ai", v_bar_term(x, z, a, i, a, a), t_2_func(x, z))
            for w in range(site):
                if (w != x) and (w != z):
                    term += np.einsum("acik, klcl -> ai", t_2_func(x, w),  v_bar_term(z, w, i, i, i, a))
    return term
```

```python
def R_bar_2_sym(x: int, y: int):
    term = v_bar_term(x, y, a, a, i, i)

    # print(term)
    term += np.einsum("abcd, cdij -> abij", v_bar_term(x, y, a, a, a, a), t_2_func(x, y))

    # print("Term_1", np.einsum("abcd, cdij -> abij", v_bar_term(x, y, a, a, a, a), t_2_func(x, y)))
    for z in range(site):
        for w in range(site):
            if z not in {x, y} and w not in {x, y} and z != w:
                term += np.einsum("klcd, acik, bdjl -> abij", v_bar_term(z, w, i, i, a, a), t_2_func(x, z), t_2_func(y, w))

    term -= t_2_func(x, y)*(E_c(x) + E_c(y) - e_bar_c(x, y))

    print(E_c(x) + E_c(y) - e_bar_c(x, y))
    return term
```

```python
def R_bar_2_non_sym(x: int, y: int):
    term = np.einsum("ac, cbij -> abij", h_bar_term(x, a, a) , t_2_func(x, y))

    # print( h_bar_term(x, a, a))

    for z in range(site):
        if z not in {x, y}:
            term += np.einsum("acik, kbcj -> abij", t_2_func(x, z), v_bar_term(z, y, i, a, a, i))
            term += np.einsum("bldl, adij -> abij", v_bar_term(y, z, a, i, a, i), t_2_func(x, y))
    return term
```

```python
def R_bar_2_total(x, y):
    return R_bar_2_sym(x, y) + R_bar_2_non_sym(x, y) + R_bar_2_non_sym(y, x)
```

```python
def R_n_term(x, y):
    x_y = np.einsum("bj, jc, acij -> abij", t_1_func(y), h_term(i, a), t_2_func(x, y))
    y_x = np.einsum("bj, jc, acij -> abij", t_1_func(x), h_term(i, a), t_2_func(y, x))
    return x_y + y_x
```
